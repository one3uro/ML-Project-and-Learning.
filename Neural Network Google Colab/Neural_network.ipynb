```python
"""
Import Libraries
"""
import pandas as pd
import os
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import PolynomialFeatures
import numpy as np
import math
```

```python
"""
Load CSV data set
"""
file_path = r"Neural Network Google Colab/creditcard.csv"
try:
    df = pd.read_csv(file_path)
    print("Successfully Loaded!")
    print("Dataset Shape:", df.shape)
    print("\nFirst 5 rows:")
    print(df.head())
except FileNotFoundError:
    print(f"Error: Could not find '{file_path}' in the working directory.")
    exit()
```

```python
"""
Define Features
"""
X = df.drop(columns=['Class'])
y = df['Class']


#Test if csv can be read
#print(X.shape)
#print(y.shape) 
#print(df['Class'].value_counts())
```

```python
"""
Perform train test split on the defined features
"""
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    random_state=42,
    test_size=0.2,
    stratify=y
)
```

```python
"""
Scale Feature Values.
"""
scaled_features = StandardScaler()
X_Scaled_train = scaled_features.fit_transform(X_train)
X_Scaled_test = scaled_features.transform(X_test)
```

```python
"""
Load model
"""
#Neural Netowrk starts here,
#We define hidden layers so here it would be 1 hidden layer with 10 neurons
model = MLPClassifier(
    hidden_layer_sizes=(10,),
    activation='relu',
    max_iter=1000,
    random_state=42
)

model.fit(X_Scaled_train, y_train)
print("Training iterations:", model.n_iter_)
print("Converged within iteration limit:", model.n_iter_ < model.max_iter)

prediction = model.predict(X_Scaled_test)
Accuracy = accuracy_score(y_test, prediction)

print(classification_report(y_test, prediction))
```
